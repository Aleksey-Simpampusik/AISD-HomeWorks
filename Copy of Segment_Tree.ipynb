# Зачем нам вообще это

Пусть есть массив:

```text
[2, 5, 1, 4, 3, 7, 6, 8]
```

Нужно поддерживать две операции:

```text
SUM L R — получить сумму на [L..R]
SET i x — заменить a[i] на x
```

Обычный проход:

```text
SUM → O(n)
```

Prefix Sum:

```text
SUM → O(1)
SET → O(n)
```

Но теперь операции постоянно чередуются:

```text
SET
SUM
SET
SUM
...
```

Нам нужна структура, в которой **и запрос, и изменение работают быстро**

# Идея структуры

Мы заранее храним агрегаты для разных кусочков массива:

```text
                        [0..7] = 36
                 /                      \
          [0..3] = 12                [4..7] = 24
          /       \                   /       \
     [0..1]=7   [2..3]=5          [4..5]=10  [6..7]=14
       / \        / \              / \        / \
      2   5       1   4            3   7       6   8
```

Каждая вершина отвечает за свой отрезок

Для суммы значение внутренней вершины получается объединением двух детей:

```text
left + right
```

Ожидаемые сложности:

```text
build  → O(n)
query  → O(log n)
update → O(log n)
```

# Как дерево хранится в памяти

Отдельные объекты `Node` нам не нужны. Мы можем хранить дерево в обычном списке:

```python
tree = [0] * (4 * n)
```

Возьмем некоторое соглашение:

```text
v              — номер текущей вершины
2 * v          — левый ребенок
2 * v + 1      — правый ребенок
```

Кроме номера вершины рекурсивным методам нужно знать, **за какой отрезок она отвечает** (надеюсь это дял вас очевидно):

```text
tl — левая граница сегмента
tr — правая граница сегмента
```

# Что должен делать `_build`

Метод получает:

```text

```

и должен сделать так, чтобы:

```text

```

Ваша цель:

> После завершения `_build(1, 0, n - 1)` корень должен хранить сумму всего массива, а остальные вершины — суммы своих сегментов

# Что должен делать `query`

Внешний метод вызывается просто:

```python
st.query(l, r)
```

Его задача небольшая:

- проверить корректность границ
- обработать пустой массив, если вы хотите его поддерживать
- запустить внутренний `_query` **с корня дерева**

Сам поиск ответа происходит в `_query`

# Что должен делать `_query`

Текущая вершина отвечает за `[tl..tr]`, а нас интересует `[l..r]`

Для каждой вершины нужно определить один из трех случаев:

### 1. Текущий сегмент полностью снаружи запроса

Он ничего не добавляет к сумме

Подумайте, какое значение можно вернуть, чтобы оно **не изменило результат сложения**

### 2. Текущий сегмент полностью внутри запроса

В `tree[v]` уже лежит готовый ответ

Глубже идти не нужно

### 3. Сегменты пересекаются частично

Взять `tree[v]` целиком нельзя

Нужно обратиться к дочерним сегментам и объединить их результаты

# Что должен делать `update`

Внешний вызов:

```python
st.update(pos, value)
```

должен:

- проверить позицию
- при необходимости обновить сохранённый исходный массив
- запустить внутренний `_update` от корня

Само движение по дереву выполняет `_update`

# Что должен делать `_update`

Изменяется только один элемент массива.

Подумайте:

- В каком листе дерева находится позиция `pos`?
- Как на каждом уровне понять, идти в левого или правого ребенка?
- Что изменить, когда дошли до нужного листа?
- Какие вершины после этого стали неактуальными?
- Как пересчитать вершину после возвращения из рекурсии?
- Нужно ли вообще заходить в соседнюю ветку дерева?

Главная идея:

> Меняем один лист и пересчитываем только его предков до корня

# Теперь соберите весь класс

Не пытайтесь сразу написать все идеально. Удобный порядок:

```text
1. __init__ + _build
2. query + _query
3. update + _update
4. тесты
```

````python
class SegmentTree:
    def __init__(self, a):
        self.a = list(a)
        self.n = len(a)
        self.tree = [0] * (4 * self.n)

        if self.n > 0:
            self._build(1, 0, self.n - 1)

    def _build(self, v, tl, tr):
        if tl == tr:
            self.tree[v] = self.a[tl]
            return

        tm = (tl + tr) // 2

        self._build(2 * v, tl, tm)
        self._build(2 * v + 1, tm + 1, tr)

        self.tree[v] = self.tree[2 * v] + self.tree[2 * v + 1]

    def query(self, l, r):
        if self.n == 0:
            raise ValueError("Массив пуст")

        if l < 0 or r >= self.n or l > r:
            raise IndexError("Некорректные границы")

        return self._query(1, 0, self.n - 1, l, r)

    def _query(self, v, tl, tr, l, r):
        if l > tr or r < tl:
            return 0

        if l <= tl and tr <= r:
            return self.tree[v]

        tm = (tl + tr) // 2

        left = self._query(2 * v, tl, tm, l, r)
        right = self._query(2 * v + 1, tm + 1, tr, l, r)

        return left + right

    def update(self, pos, value):
        if pos < 0 or pos >= self.n:
            raise IndexError("Некорректная позиция")

        self.a[pos] = value
        self._update(1, 0, self.n - 1, pos, value)

    def _update(self, v, tl, tr, pos, value):
        if tl == tr:
            self.tree[v] = value
            return

        tm = (tl + tr) // 2

        if pos <= tm:
            self._update(2 * v, tl, tm, pos, value)
        else:
            self._update(2 * v + 1, tm + 1, tr, pos, value)

        self.tree[v] = self.tree[2 * v] + self.tree[2 * v + 1]
````

# Тест 1. Построение

Если `build` работает правильно, корень для этого массива должен хранить сумму всего массива

````python
a = [2, 5, 1, 4, 3, 7, 6, 8]
st = SegmentTree(a)

assert st.tree[1] == 36
print("BUILD")
````

# Тест 2. Запросы

````python
a = [2, 5, 1, 4, 3, 7, 6, 8]
st = SegmentTree(a)

tests = [
    (0, 0, 2),
    (0, 3, 12),
    (2, 6, 21),
    (4, 7, 24),
    (0, 7, 36),
    (7, 7, 8),
]

for l, r, expected in tests:
    actual = st.query(l, r)
    assert actual == expected, (
        f"query({l}, {r}) → {actual}, ожидалось {expected}"
    )

print("QUERY")
````

# Тест 3. Обновление

После:

# Тест 4. Несколько изменений подряд

Segment Tree нужен именно для сценария:

```text
update → query → update → query
```

````python
a = [2, 5, 1, 4, 3, 7, 6, 8]
st = SegmentTree(a)

assert st.query(2, 6) == 21

st.update(3, 10)
assert st.query(2, 6) == 27

st.update(5, 1)
assert st.query(2, 6) == 21

st.update(0, 100)
assert st.query(0, 1) == 105

print("УРА МЫ КРУТЫЕ")
````

##**Интеграция Segment Tree в индивидуальный проект**
Шаг 1. Выбор последовательности
**Поле, задающее порядок записей:** Order Date — дата оформления заказа.

**Числовое поле, которое может изменяться:** Total Profit — прибыль от заказа.

Записи сортируются по дате оформления заказа. При одинаковой дате сохраняется исходный порядок.

##**Шаг 2. Динамический сценарий**


**Сценарий:** система анализа прибыли от продаж, в которой могут корректироваться финансовые показатели отдельных заказов. Необходимо быстро рассчитывать суммарную прибыль за диапазон записей, отсортированных по дате.

**Операция UPDATE i x:** изменить прибыль заказа с индексом i на новое значение x.

**Операция QUERY L R:** получить суммарную прибыль заказов с индекса L по индекс R включительно.

Для обеих операций используется нумерация с нуля.

Segment Tree подходит для этой задачи, поскольку позволяет выполнять изменение отдельного значения и вычисление суммы диапазона за O(log N).

##**Шаг 3. Реализация индекса**

````python

import pandas as pd
import time
import random

df = pd.read_csv("sample_data/100000 Sales Records.csv")

df["Order Date"] = pd.to_datetime(
    df["Order Date"], format="%m/%d/%Y"
)

df = df.sort_values(
    "Order Date", kind="stable"
).reset_index(drop=True)

data = df["Total Profit"].astype(float).tolist()


start = time.perf_counter()
index = SegmentTree(data)
build_time = time.perf_counter() - start

print("Время построения:", build_time)

print("Прибыль первых 100 заказов:",
      index.query(0, 99))

index.update(10, 50000)

print("После UPDATE:",
      index.query(0, 99))
````

##**Шаг 4. Сравнение с наивным решением**
Создадим 2 000 операций, из которых 1 000 — обновления и 1 000 — запросы сумм.

Одинаковая последовательность операций будет выполнена на обычном массиве и на Segment Tree.

````python

rng = random.Random(42)

n = len(data)
operations = []

for k in range(2000):
    if k % 2 == 0:
        i = rng.randrange(n)
        x = round(data[i] * rng.uniform(0.8, 1.2), 2)
        operations.append(("update", i, x))
    else:
        l = rng.randrange(n)
        r = rng.randrange(l, n)
        operations.append(("query", l, r))


naive = data.copy()
naive_answers = []

start = time.perf_counter()

for op, i, x in operations:
    if op == "update":
        naive[i] = x
    else:
        naive_answers.append(sum(naive[i:x + 1]))

naive_time = time.perf_counter() - start


start = time.perf_counter()
index = SegmentTree(data)
build_time = time.perf_counter() - start

tree_answers = []

start = time.perf_counter()

for op, i, x in operations:
    if op == "update":
        index.update(i, x)
    else:
        tree_answers.append(index.query(i, x))

tree_time = time.perf_counter() - start


correct = all(
    abs(a - b) <= 1e-7 * max(1, abs(a), abs(b))
    for a, b in zip(naive_answers, tree_answers)
)

print("Количество update:", 1000)
print("Количество query:", 1000)
print("Время построения:", build_time)
print("Обычный массив:", naive_time)
print("Segment Tree:", tree_time)
print("Ускорение:", naive_time / tree_time)
print("Результаты совпадают:", correct)
````
